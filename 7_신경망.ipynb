# 인공 신경망 (Artificial Neural Network)

## 인공 신경망이란?

인공 신경망(Artificial Neural Network, ANN)은 인간의 뇌 신경세포(뉴런)의 구조를 모방하여 만든 머신러닝 모델이다.

입력 데이터를 여러 층(Layer)을 거치면서 처리하고, 가중치와 편향을 학습하여 원하는 결과를 예측한다.

---

## 인공 신경망의 구조

인공 신경망은 크게 세 종류의 층으로 구성된다.

### 입력층 (Input Layer)

입력 데이터를 받아들이는 층이다.

예를 들어 이미지가 28×28 픽셀이라면 총 784개의 입력값이 입력층으로 들어간다.

### 은닉층 (Hidden Layer)

입력 데이터로부터 중요한 특징을 추출하는 층이다.

은닉층이 많아질수록 복잡한 패턴을 학습할 수 있다.

### 출력층 (Output Layer)

최종 결과를 출력하는 층이다.

분류 문제에서는 클래스 개수만큼 뉴런을 가진다.

---

## 뉴런의 계산

하나의 뉴런은 다음과 같은 계산을 수행한다.

$$
z = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b
$$

여기서

- $x$ : 입력값
- $w$ : 가중치(Weight)
- $b$ : 편향(Bias)

이다.

행렬 형태로 표현하면

$$
Z = XW + b
$$

가 된다.

---

## 활성화 함수

선형 계산만 수행하면 복잡한 문제를 해결할 수 없기 때문에 활성화 함수를 사용한다.

### Sigmoid

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

출력 범위는 0~1이다.

### ReLU

$$
ReLU(x)=\max(0,x)
$$

현재 가장 많이 사용되는 활성화 함수이다.

### Softmax

$$
P(y_i)=\frac{e^{z_i}}{\sum_j e^{z_j}}
$$

출력값을 확률로 변환한다.

---

## 딥러닝

은닉층이 여러 개인 신경망을 딥러닝(Deep Learning)이라고 한다.

은닉층이 깊어질수록 더욱 복잡한 패턴을 학습할 수 있다.

---

# 텐서플로 (TensorFlow)

## 텐서플로란?

TensorFlow는 구글이 개발한 오픈소스 딥러닝 프레임워크이다.

복잡한 신경망을 쉽게 구현하고 학습할 수 있도록 다양한 기능을 제공한다.

현재는 Keras가 포함되어 있어 간단한 코드로 신경망을 만들 수 있다.

---

## 텐서(Tensor)

TensorFlow에서 사용하는 기본 데이터 구조이다.

### 0차원 텐서 (스칼라)

하나의 값

$$
5
$$

### 1차원 텐서 (벡터)

$$
[1,2,3]
$$

### 2차원 텐서 (행렬)

$$
\begin{bmatrix}

\end{bmatrix}
$$

### 3차원 이상 텐서

예를 들어

$$
(300,100,100)
$$

은

- 데이터 300개
- 이미지 높이 100

을 의미한다.

---

## TensorFlow의 장점

# 밀집층 (Dense Layer)

## 밀집층이란?

Dense Layer는 이전 층의 모든 뉴런과 현재 층의 모든 뉴런이 연결된 층이다.

완전 연결층(Fully Connected Layer)이라고도 부른다.

---

## Dense Layer의 계산

입력 행렬을

$$
X
$$

가중치 행렬을

$$
W
$$

편향을

$$
b
$$

라고 하면 출력은

$$
Y = XW + b
$$

로 계산된다.

---

## Dense Layer의 역할

- 데이터의 특징 학습
- 입력과 출력 사이의 관계 학습
- 분류 및 회귀 문제 해결

---

## Dense Layer의 뉴런 수

예를 들어

text Dense(10)

은 출력 뉴런이 10개라는 의미이다.

즉, 입력이 무엇이든 최종 출력은 10개의 값으로 구성된다.

---

## Dense Layer와 활성화 함수

Dense Layer 뒤에는 일반적으로 활성화 함수를 연결한다.

예시

text Dense → ReLU Dense → Sigmoid Dense → Softmax

---

# 원-핫 인코딩 (One-Hot Encoding)

## 원-핫 인코딩이란?

범주형 데이터를 신경망이 학습할 수 있는 형태의 벡터로 변환하는 방법이다.

---

## 예시

과일 분류 문제

| 클래스 | 숫자 |
|---------|---------|
| 사과 | 0 |
| 바나나 | 1 |
| 파인애플 | 2 |

원-핫 인코딩을 적용하면

### 사과

$$
[1,0,0]
$$

### 바나나

$$
[0,1,0]
$$

### 파인애플

$$
[0,0,1]
$$

가 된다.

---

## 왜 사용하는가?

만약

text 사과 = 0 바나나 = 1 파인애플 = 2

로 표현하면

신경망은

$$
2 > 1 > 0
$$

이라는 크기 관계가 있다고 잘못 이해할 수 있다.

원-핫 인코딩은 이러한 문제를 해결한다.

---

## 출력층과 원-핫 인코딩

클래스가 3개인 경우 출력층의 결과가

$$
[0.1,\ 0.8,\ 0.1]
$$

이라고 하자.

실제 정답이 바나나라면

$$
[0,\ 1,\ 0]
$$

으로 표현한다.

이 두 벡터를 비교하여 손실 함수를 계산하고 가중치를 업데이트한다.

---

## 원-핫 인코딩의 장점

- 클래스 간 크기 관계 제거
- 다중 분류 문제에 적합
- Softmax와 함께 사용 가능
- 교차 엔트로피 손실 함수와 궁합이 좋음

---

# 정리

- 인공 신경망은 인간의 뉴런 구조를 모방한 머신러닝 모델이다.
- TensorFlow는 딥러닝 모델을 쉽게 구현할 수 있는 프레임워크이다.
- Dense Layer는 모든 뉴런이 서로 연결된 완전 연결층이다.
- One-Hot Encoding은 범주형 데이터를 벡터 형태로 변환하는 방법이다.
- 다중 분류 문제에서는 일반적으로 Softmax + One-Hot Encoding + 교차 엔트로피 손실 함수를 함께 사용한다.

`````python
from tensorflow import keras
(train_input, train_target), (test_input, test_target) = keras.datasets.fashion_mnist.load_data()

#전달받은 데이터의 크기 확인
print(train_input.shape, train_target.shape)
print(test_input.shape, test_target.shape)
`````

`````python
#전달받은 데이터 중 몇몇 출력
import matplotlib.pyplot as plt
fig, axs = plt.subplots(1, 10, figsize=(10,10))
for i in range(10):
	axs[i].imshow(train_input[i], cmap='gray_r')
	axs[i].axis('off')
plt.show()
`````

`````python
print([train_target[i] for i in range(10)])
`````

`````python
#데이터의 종류와 갯수 출력
import numpy as np
print(np.unique(train_target, return_counts=True))


#mnist의 경우 0~255 사이의 정수값을 가지므로, 255로 나누어 0~1 사이의 값으로 정규화한다.

train_scaled = train_input / 255.0
train_scaled = train_scaled.reshape(-1, 28*28)

print(train_scaled.shape)
`````

`````python
#SGDClassifier의 개념, 파라미터에 대한 설명
from sklearn.model_selection import cross_validate
from sklearn.linear_model import SGDClassifier
sc = SGDClassifier(loss='log_loss', max_iter=5, random_state=42)
scores = cross_validate(sc, train_scaled, train_target, n_jobs=-1)
print(np.mean(scores['test_score']))
`````

# SGDClassifier

## SGDClassifier란?

SGDClassifier는 확률적 경사 하강법(Stochastic Gradient Descent, SGD) 을 사용하여 분류 문제를 해결하는 모델이다.

데이터를 한 번에 모두 사용하는 것이 아니라 하나씩 또는 작은 묶음 단위로 학습하며 가중치를 업데이트한다.

대용량 데이터에서도 빠르게 학습할 수 있다는 장점이 있다.

---

## SGDClassifier 파라미터

| 파라미터 | 값 | 설명 |
|-----------|-----------|-----------|
| loss | 'log_loss' | 로지스틱 회귀 방식으로 학습 |
| max_iter | 5 | 전체 훈련 데이터를 5번 반복 학습 |
| random_state | 42 | 난수 시드를 고정하여 항상 같은 결과 생성 |

---

## cross_validate() 파라미터

| 파라미터 | 값 | 설명 |
|-----------|-----------|-----------|
| 첫 번째 인수 | sc | 평가할 모델 |
| 두 번째 인수 | train_scaled | 훈련 데이터 |
| 세 번째 인수 | train_target | 정답 데이터 |
| n_jobs | -1 | 모든 CPU 코어를 사용하여 병렬 처리 |

---

## cross_validate()

교차 검증(Cross Validation)을 수행하는 함수이다.

기본적으로 5-Fold 교차 검증을 수행하며, 데이터를 5개로 나누어 학습과 검증을 반복한다.

이를 통해 모델의 일반화 성능을 더 정확하게 평가할 수 있다.

---

## scores['test_score']

각 Fold에서 측정한 검증 정확도가 저장된다.

예시

python scores['test_score']

출력

python [0.81, 0.83, 0.82, 0.84, 0.80]

---

## np.mean(scores['test_score'])

교차 검증에서 얻은 정확도의 평균을 계산한다.

python print(np.mean(scores['test_score']))

예시 결과

python 0.82

즉, 모델의 평균 검증 정확도가 82%라는 의미이다.

---

## 실행 순서

1. SGDClassifier 모델 생성
2. 교차 검증 수행
3. 각 Fold의 정확도 저장
4. 평균 정확도 계산
5. 모델 성능 확인

`````python
import tensorflow as tf
from tensorflow import keras

from sklearn.model_selection import train_test_split
train_scaled, val_scaled, train_target, val_target = train_test_split(train_scaled, train_target, test_size=0.2, random_state=42)

#데이터가 train데이터와 validate데이터로 잘 분할되었는지 확인
print(train_scaled.shape, train_target.shape)
print(val_scaled.shape, val_target.shape)
`````

`````python
dense = keras.layers.Dense(10, activation='softmax', input_shape=(784,))
model = keras.Sequential([dense])
model.compile(loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(train_scaled, train_target, epochs=5)
model.evaluate(val_scaled, val_target)
`````

```text
      ↓

Softmax

      ↓

10개 클래스 확률 출력
```

이 모델은 **은닉층(Hidden Layer)이 없는 가장 간단한 신경망**이며, 사실상 **다중 클래스 로지스틱 회귀(Logistic Regression)** 와 동일한 구조이다.

# 심층 신경망(Deep Neural Network)

심층 신경망(Deep Neural Network, DNN)은 입력층과 출력층 사이에 여러 개의 은닉층(Hidden Layer)을 가진 인공 신경망이다.

은닉층이 많아질수록 데이터의 복잡한 패턴을 학습할 수 있다.

---

# 신경망의 구조

## 1. 단층 신경망

```text
입력층
   ↓
출력층
```

은닉층이 없는 가장 단순한 형태이다.

---

## 2. 심층 신경망

```text
입력층
   ↓
은닉층 1
   ↓
은닉층 2
   ↓
은닉층 3
   ↓
출력층
```

은닉층이 여러 개 존재한다.

---

# 심층 신경망의 장점

- 복잡한 데이터 학습 가능
- 이미지 인식 성능 향상
- 음성 인식 성능 향상
- 자연어 처리 성능 향상

---

# 심층 신경망의 학습 과정

## 1단계 : 순전파(Forward Propagation)

입력 데이터를 각 층에 전달한다.

```text
입력
 ↓
은닉층
 ↓
출력층
```

---

## 2단계 : 손실 계산

예측값과 정답을 비교하여 오차를 계산한다.

---

## 3단계 : 역전파(Backpropagation)

오차를 출력층에서 입력층 방향으로 전달한다.

---

## 4단계 : 가중치 업데이트

옵티마이저를 이용하여 가중치를 수정한다.

---

# 활성화 함수(Activation Function)

신경망은 선형 계산만 수행하면 복잡한 문제를 학습할 수 없다.

따라서 비선형성을 추가하기 위해 활성화 함수를 사용한다.

---

# ReLU 함수

ReLU(Rectified Linear Unit)는 현재 가장 널리 사용되는 활성화 함수이다.

수식은 다음과 같다.

$$
ReLU(x)
=
\max(0,x)
$$

---

## 동작 원리

### 입력이 양수

$$
x > 0
$$

이면

$$
ReLU(x)=x
$$

---

### 입력이 음수

$$
x < 0
$$

이면

$$
ReLU(x)=0
$$

---

## 예시

| 입력 | 출력 |
|--------|--------|
| -5 | 0 |
| -2 | 0 |
| 0 | 0 |
| 3 | 3 |
| 10 | 10 |

---

## 그래프 형태

```text
출력
 ^
 |
 |       /
 |      /
 |     /
 |____/________→ 입력
      0
```

---

## ReLU의 장점

- 계산 속도가 빠름
- 기울기 소실 문제 감소
- 딥러닝에서 높은 성능

---

## ReLU의 단점

입력이 계속 음수이면

$$
ReLU(x)=0
$$

이 되어 학습이 멈출 수 있다.

이를 Dead ReLU 문제라고 한다.

---

# 옵티마이저(Optimizer)

옵티마이저는 손실 함수를 최소화하도록 가중치를 업데이트하는 알고리즘이다.

---

# SGD (Stochastic Gradient Descent)

확률적 경사 하강법이다.

가장 기본적인 옵티마이저이다.

---

## 가중치 업데이트

$$
W = W - \eta \nabla L
$$

여기서

- $W$ : 가중치
- $\eta$ : 학습률(Learning Rate)
- $\nabla L$ : 손실 함수의 기울기

---

## 특징

### 장점

- 구현이 간단
- 메모리 사용량 적음

### 단점

- 수렴 속도가 느림
- 지역 최적점에 빠질 수 있음

---

# 모멘텀(Momentum)

모멘텀은 이전 기울기를 기억하여 더 빠르게 이동하는 방법이다.

공이 굴러가는 관성을 이용하는 것과 비슷하다.

---

## 특징

- SGD보다 빠르게 수렴
- 진동 감소
- 지역 최적점 탈출 도움

---

# 네스테로프 모멘텀 (Nesterov Momentum)

Nesterov Accelerated Gradient(NAG)라고도 한다.

---

## 아이디어

일반 모멘텀은 현재 위치에서 기울기를 계산한다.

네스테로프 모멘텀은

"앞으로 이동할 위치를 먼저 예측"

한 뒤 그 위치에서 기울기를 계산한다.

---

## 특징

- 일반 모멘텀보다 정확함
- 더 빠르게 수렴
- 진동 감소

---

## 비교

### SGD

```text
현재 위치
   ↓
기울기 계산
   ↓
이동
```

---

### Nesterov

```text
이동 예상
   ↓
예상 위치에서 기울기 계산
   ↓
업데이트
```

---

# RMSprop

RMSprop(Root Mean Square Propagation)은 학습률을 자동으로 조절하는 옵티마이저이다.

---

## 문제점

SGD는 모든 가중치에 동일한 학습률을 사용한다.

하지만 어떤 가중치는 크게 수정해야 하고,

어떤 가중치는 조금만 수정해야 할 수 있다.

---

## 해결 방법

각 가중치마다 다른 학습률을 적용한다.

최근 기울기의 제곱 평균을 사용하여 학습률을 조절한다.

---

## 특징

### 장점

- 빠른 수렴
- 학습률 자동 조정
- 딥러닝에서 좋은 성능

### 단점

- 하이퍼파라미터 설정 필요

---

# Adam

Adam(Adaptive Moment Estimation)은 현재 가장 널리 사용되는 옵티마이저이다.

---

## 핵심 아이디어

Adam은

- Momentum
- RMSprop

을 결합한 알고리즘이다.

---

## Adam이 사용하는 정보

### 1차 모멘트

기울기의 평균

$$
m_t
$$

---

### 2차 모멘트

기울기 제곱의 평균

$$
v_t
$$

---

이를 이용해 가중치를 업데이트한다.

---

# Adam의 특징

### 장점

- 매우 빠른 수렴
- 안정적인 학습
- 학습률 자동 조정
- 대부분의 문제에서 좋은 성능

### 단점

- 메모리 사용량 증가
- 경우에 따라 일반 SGD보다 일반화 성능이 떨어질 수 있음

---

# 옵티마이저 비교

| 옵티마이저 | 특징 |
|------------|--------|
| SGD | 가장 기본적인 경사 하강법 |
| Momentum | 관성을 이용해 빠르게 이동 |
| Nesterov Momentum | 미래 위치를 예측하여 업데이트 |
| RMSprop | 학습률을 자동 조정 |
| Adam | Momentum + RMSprop |

---

# 일반적인 사용 방법

| 상황 | 추천 |
|--------|--------|
| 딥러닝 입문 | Adam |
| 빠른 실험 | Adam |
| CNN 학습 | Adam |
| Transformer 학습 | Adam |
| 최고 성능 튜닝 | SGD + Momentum |

---

# 손실 곡선(Loss Curve)

손실 곡선(Loss Curve)은 학습 과정에서 손실 함수(Loss Function)의 값이 어떻게 변화하는지를 그래프로 나타낸 것이다.

신경망이 학습을 잘 하고 있는지 확인하는 가장 중요한 방법 중 하나이다.

---

# 손실 함수(Loss Function)

손실 함수는 모델의 예측값과 실제 정답 사이의 차이를 수치로 나타낸 값이다.

$$
Loss = \text{예측값과 정답의 오차}
$$

손실값이 작을수록 모델의 성능이 좋다고 볼 수 있다.

---

# 손실 곡선의 의미

학습이 진행될수록 손실값은 감소해야 한다.

```text
Loss
 ^
 |
 |\
 | \
 |  \
 |   \
 |    \____
 |
 +-----------------> Epoch
```

위 그래프처럼 손실이 꾸준히 감소하면 정상적으로 학습되고 있는 것이다.

---

# 손실 곡선으로 알 수 있는 것

## 정상 학습

```text
Loss
 ^
 |
 |\
 | \
 |  \
 |   \
 |    \____
 |
 +-----------------> Epoch
```

손실이 점점 감소한다.

---

## 과소적합(Underfitting)

```text
Loss
 ^
 |
 |---------
 |
 +-----------------> Epoch
```

손실이 거의 감소하지 않는다.

모델이 충분히 학습하지 못한 상태이다.

---

## 과적합(Overfitting)

훈련 손실은 감소하지만 검증 손실은 증가한다.

```text
Loss
 ^
 |\
 | \
 |  \
 |   \
 |    \__
 |       \__
 |          \__
 +-----------------> Epoch
```

---

# 검증 손실(Validation Loss)

검증 손실은 검증 데이터에 대한 손실값이다.

---

## 훈련 손실

훈련 데이터로 계산한 손실

$$
Training\ Loss
$$

---

## 검증 손실

검증 데이터로 계산한 손실

$$
Validation\ Loss
$$

---

# 훈련 손실과 검증 손실 비교

## 정상 학습

```text
Loss
 ^
 |\
 | \
 |  \
 |   \
 |    \__
 |
 +-----------------> Epoch
```

훈련 손실과 검증 손실이 함께 감소한다.

---

## 과적합 발생

```text
Loss
 ^
 |\
 | \
 |  \
 |   \____
 |        \____
 |
 |     /\
 |    /  \
 |   /    \
 +-----------------> Epoch
```

- 훈련 손실 : 계속 감소
- 검증 손실 : 다시 증가

검증 손실이 증가하기 시작하면 과적합이 발생한 것이다.

---

# 드롭아웃(Dropout)

드롭아웃은 과적합을 방지하기 위한 대표적인 규제(Regularization) 방법이다.

---

# 드롭아웃의 아이디어

학습할 때 일부 뉴런을 무작위로 제거한다.

예를 들어

```text
입력층
 ↓
○ ○ ○ ○ ○
 ↓
은닉층
 ↓
출력층
```

---

학습 중

```text
입력층
 ↓
○ X ○ X ○
 ↓
은닉층
 ↓
출력층
```

일부 뉴런을 비활성화한다.

---

# 드롭아웃 비율

예를 들어

```python
Dropout(0.3)
```

이라면

30%의 뉴런을 무작위로 제거한다.

---

# 드롭아웃의 장점

- 과적합 감소
- 일반화 성능 향상
- 특정 뉴런에 대한 의존성 감소

---

# 드롭아웃의 동작

## 학습 시

드롭아웃 적용

```text
일부 뉴런 제거
```

---

## 예측 시

드롭아웃 적용 안 함

```text
모든 뉴런 사용
```

---

# 콜백(Callback)

콜백은 학습 중 특정 시점에 자동으로 실행되는 기능이다.

---

# 콜백의 목적

학습 중에

- 모델 저장
- 조기 종료
- 학습률 변경

등을 자동으로 수행한다.

---

# 콜백 동작 과정

```text
Epoch 1 종료
 ↓
콜백 실행

Epoch 2 종료
 ↓
콜백 실행

Epoch 3 종료
 ↓
콜백 실행
```

---

# 대표적인 콜백

## ModelCheckpoint

최고 성능 모델 저장

---

## EarlyStopping

조기 종료 수행

---

## ReduceLROnPlateau

학습률 자동 감소

---

# 조기 종료(Early Stopping)

조기 종료는 과적합이 발생하기 전에 학습을 중단하는 기법이다.

---

# 조기 종료가 필요한 이유

학습을 오래 하면

- 훈련 손실 감소
- 검증 손실 증가

현상이 발생할 수 있다.

이를 과적합이라고 한다.

---

# 조기 종료의 동작

예를 들어

| Epoch | 검증 손실 |
|---------|---------|
| 1 | 0.50 |
| 2 | 0.42 |
| 3 | 0.37 |
| 4 | 0.35 |
| 5 | 0.36 |
| 6 | 0.38 |
| 7 | 0.41 |

---

가장 좋은 성능은

```text
Epoch 4
```

이다.

따라서 이후 학습을 중단한다.

---

# patience

조기 종료는 보통 patience 값을 사용한다.

예를 들어

```python
EarlyStopping(patience=2)
```

이면

검증 손실이 2번 연속 개선되지 않으면 학습을 종료한다.

---

# 조기 종료의 장점

- 과적합 방지
- 학습 시간 단축
- 최적 모델 자동 선택

---

# 손실 곡선과 조기 종료

```text
Loss
 ^
 |
 |\
 | \
 |  \
 |   \__
 |      \__
 |
 |    \______
 |
 +-----------------> Epoch
          ↑
      Early Stopping
```

검증 손실이 증가하기 시작하는 지점에서 학습을 종료한다.

# 핵심 요약

- 손실 곡선은 학습 상태를 확인하는 그래프이다.
- 검증 손실은 모델의 일반화 성능을 평가한다.
- 드롭아웃은 과적합을 줄이는 대표적인 방법이다.
- 콜백은 학습 과정에서 자동으로 특정 작업을 수행한다.
- 조기 종료는 검증 손실이 더 이상 개선되지 않을 때 학습을 멈추어 과적합을 방지한다.

`````python
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import numpy as np
import pandas as pd


(train_input, train_target), (test_input, test_target) = keras.datasets.fashion_mnist.load_data()

train_scaled = train_input / 255.0
train_scaled, val_scaled, train_target, val_target = train_test_split(
    train_scaled, train_target, test_size=0.2, random_state=42)
`````

`````python
def model_fn(a_layer=None):
  model = keras.Sequential()
  model.add(keras.layers.Flatten(input_shape=(28, 28)))
  model.add(keras.layers.Dense(100, activation='relu'))

  if a_layer:
    model.add(a_layer)
  model.add(keras.layers.Dense(10, activation='softmax'))

  return model
`````

`````python
model = model_fn()
model.summary()
`````

`````python
model.compile(loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history = model.fit(train_scaled, train_target, epochs=5, verbose=0)
`````

`````python
print(history.history.keys())
`````

`````python
import matplotlib.pyplot as plt

plt.plot(history.history['loss'])
plt.xlabel('epoch')
plt.ylabel('loss')
plt.show()
`````

`````python
model = model_fn()
model.compile(loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(train_scaled, train_target, epochs=20, verbose=0)
plt.plot(model.history.history['loss'])
plt.xlabel('epoch')
plt.ylabel('loss')
plt.show()
`````

`````python
# 드롭 아웃
model = model_fn(keras.layers.Dropout(0.3))
model.summary()
`````
